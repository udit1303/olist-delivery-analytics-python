```python
# Convert date columns to datetime
date_cols = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_cols:
    orders[col] = pd.to_datetime(orders[col])

# Overall date range
print("Purchase date range:")
print("Earliest:", orders["order_purchase_timestamp"].min())
print("Latest:  ", orders["order_purchase_timestamp"].max())

# Monthly order count
orders["purchase_month"] = orders["order_purchase_timestamp"].dt.to_period("M")

monthly_orders = (
    orders.groupby("purchase_month")["order_id"]
    .nunique()
)

print("\nMonthly order count:")
print(monthly_orders)
```

```python
# Order status distribution
print("ORDER STATUS:")
print(orders["order_status"].value_counts())

# Check delivery-date availability by status
print("\nDELIVERY DATE AVAILABILITY BY STATUS:")
print(
    orders.groupby("order_status")[
        ["order_delivered_carrier_date",
         "order_delivered_customer_date",
         "order_estimated_delivery_date"]
    ]
    .count()
)

# Check for impossible delivery dates
print("\nIMPOSSIBLE DATE CHECKS:")

check_1 = (
    orders["order_delivered_customer_date"]
    < orders["order_purchase_timestamp"]
).sum()

check_2 = (
    orders["order_delivered_customer_date"]
    < orders["order_delivered_carrier_date"]
).sum()

print("Delivered before purchase:", check_1)
print("Delivered to customer before carrier:", check_2)
```

```python
# Create the dataset for delivery analysis
delivered_orders = orders[
    (orders['order_status'] == 'delivered') &
    (orders['order_delivered_customer_date'].notna()) &
    (orders['order_estimated_delivery_date'].notna())
].copy()

print("Usable delivered orders:", len(delivered_orders))
```

```python
# Calculate delay in days
delivered_orders['delay_days'] = (
    delivered_orders['order_delivered_customer_date']
    - delivered_orders['order_estimated_delivery_date']
).dt.days

# Keep only late orders
late_orders = delivered_orders[
    delivered_orders['delay_days'] > 0
].copy()

# Create severity buckets
def delay_bucket(days):
    if days <= 2:
        return '1-2 days'
    elif days <= 5:
        return '3-5 days'
    elif days <= 10:
        return '6-10 days'
    else:
        return '>10 days'

late_orders['delay_bucket'] = late_orders['delay_days'].apply(delay_bucket)

# Summary
delay_summary = (
    late_orders['delay_bucket']
    .value_counts()
    .reindex(['1-2 days', '3-5 days', '6-10 days', '>10 days'])
    .reset_index()
)

delay_summary.columns = ['Delay Bucket', 'Orders']

delay_summary['% of Late Orders'] = (
    delay_summary['Orders'] / len(late_orders) * 100
).round(2)

print(delay_summary)

print("\nAverage days late:", round(late_orders['delay_days'].mean(), 2))
print("Median days late:", round(late_orders['delay_days'].median(), 2))
print("Maximum days late:", late_orders['delay_days'].max())
```

```python
# Merge customer state into our delivery dataset
delivery_geo = delivered_orders.merge(
    customers[['customer_id', 'customer_state']],
    on='customer_id',
    how='left'
)

# Mark late orders
delivery_geo['is_late'] = delivery_geo['delay_days'] > 0

# State-level delivery performance
state_delivery = (
    delivery_geo.groupby('customer_state')
    .agg(
        Orders=('order_id', 'count'),
        Late_Orders=('is_late', 'sum'),
        Avg_Delivery_Days=(
            'order_delivered_customer_date',
            lambda x: (x - delivery_geo.loc[x.index, 'order_purchase_timestamp']).dt.total_seconds().mean() / 86400
        ),
        Avg_Delay_Days=('delay_days', 'mean')
    )
    .reset_index()
)

# Calculate late %
state_delivery['Late_%'] = (
    state_delivery['Late_Orders'] / state_delivery['Orders'] * 100
).round(2)

# Round other metrics
state_delivery['Avg_Delivery_Days'] = state_delivery['Avg_Delivery_Days'].round(2)
state_delivery['Avg_Delay_Days'] = state_delivery['Avg_Delay_Days'].round(2)

# Sort by number of late orders
state_delivery = state_delivery.sort_values(
    'Late_Orders',
    ascending=False
)

print(state_delivery)
```

```python
# Look only at states with at least 500 delivered orders
state_delivery_500 = state_delivery[
    state_delivery['Orders'] >= 500
].sort_values(
    'Late_%',
    ascending=False
)

print(state_delivery_500)
```

```python
# Average delay ONLY among late orders
state_late = (
    delivery_geo[delivery_geo['is_late']]
    .groupby('customer_state')
    .agg(
        Late_Orders=('order_id', 'count'),
        Avg_Days_Late=('delay_days', 'mean'),
        Median_Days_Late=('delay_days', 'median'),
        Max_Days_Late=('delay_days', 'max')
    )
    .reset_index()
)

# Add total orders and late %
state_late = state_late.merge(
    state_delivery[['customer_state', 'Orders', 'Late_%']],
    on='customer_state',
    how='left'
)

# % of all late orders coming from each state
total_late_orders = state_late['Late_Orders'].sum()

state_late['Share_of_All_Late_Orders_%'] = (
    state_late['Late_Orders'] / total_late_orders * 100
).round(2)

# Round metrics
state_late['Avg_Days_Late'] = state_late['Avg_Days_Late'].round(2)
state_late['Median_Days_Late'] = state_late['Median_Days_Late'].round(2)

# Sort by number of late orders
state_late = state_late.sort_values(
    'Late_Orders',
    ascending=False
)

print(state_late)
```

```python
# States with meaningful volume
state_late_500 = state_late[
    state_late['Orders'] >= 500
].copy()

# Sort by late-delivery rate
print(
    state_late_500.sort_values(
        'Late_%',
        ascending=False
    )[
        [
            'customer_state',
            'Orders',
            'Late_Orders',
            'Late_%',
            'Share_of_All_Late_Orders_%',
            'Avg_Days_Late',
            'Median_Days_Late',
            'Max_Days_Late'
        ]
    ]
)
```

```python
# Connect delivered orders with order items and product information

category_orders = delivered_orders[
    ['order_id', 'delay_days', 'is_late']
].merge(
    order_items[['order_id', 'product_id']],
    on='order_id',
    how='inner'
).merge(
    products[['product_id', 'product_category_name']],
    on='product_id',
    how='left'
)

# Category-level performance
category_delivery = (
    category_orders
    .groupby('product_category_name')
    .agg(
        Orders=('order_id', 'nunique'),
        Late_Orders=('is_late', 'sum')
    )
    .reset_index()
)

# Calculate late %
category_delivery['Late_%'] = (
    category_delivery['Late_Orders'] /
    category_delivery['Orders'] * 100
).round(2)

# Share of all late orders
total_late = delivered_orders['is_late'].sum()

category_delivery['Share_of_All_Late_Orders_%'] = (
    category_delivery['Late_Orders'] /
    total_late * 100
).round(2)

# Only categories with meaningful volume
category_delivery_500 = category_delivery[
    category_delivery['Orders'] >= 500
].sort_values(
    'Late_%',
    ascending=False
)

print(category_delivery_500)
```

```python
# Create a clean order-level delivery dataset
order_delivery = delivered_orders[
    ['order_id', 'delay_days']
].copy()

order_delivery['is_late'] = order_delivery['delay_days'] > 0

# Connect orders -> items -> products
category_orders = (
    order_delivery
    .merge(
        order_items[['order_id', 'product_id']],
        on='order_id',
        how='inner'
    )
    .merge(
        products[['product_id', 'product_category_name']],
        on='product_id',
        how='left'
    )
)

# Replace missing category names
category_orders['product_category_name'] = (
    category_orders['product_category_name']
    .fillna('Unknown')
)

# One order-category combination should count only once
category_orders = category_orders.drop_duplicates(
    subset=['order_id', 'product_category_name']
)

# Category-level performance
category_delivery = (
    category_orders
    .groupby('product_category_name')
    .agg(
        Orders=('order_id', 'nunique'),
        Late_Orders=('is_late', 'sum'),
        Avg_Delay_Days=('delay_days', 'mean')
    )
    .reset_index()
)

# Late delivery %
category_delivery['Late_%'] = (
    category_delivery['Late_Orders']
    / category_delivery['Orders'] * 100
).round(2)

# Keep only categories with meaningful volume
category_delivery_500 = (
    category_delivery[
        category_delivery['Orders'] >= 500
    ]
    .sort_values('Late_%', ascending=False)
)

print(category_delivery_500)
```

```python
import numpy as np

# Haversine distance in kilometers

def haversine(lat1, lon1, lat2, lon2):
    R = 6371.0  # Earth radius in km

    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)
    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    return 2 * R * np.arcsin(np.sqrt(a))


route_data['distance_km'] = haversine(
    route_data['seller_lat'],
    route_data['seller_lng'],
    route_data['customer_lat'],
    route_data['customer_lng']
)

print(route_data['distance_km'].describe())
```
